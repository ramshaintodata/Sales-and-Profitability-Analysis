```python
# Converting data types of this dataframe
# ------------------------#
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
```

```python
# Quantifying missingness
# ------------------------#
print("Missing values:\n", df.isnull().sum())
```

```python
# Imputing missing values in brand, price, payment_method, delivery_days, region, customer_age_group, device_type
# ------------------------#
# Type: Numeric: median
df['price'].fillna(df['price'].median(), inplace=True)
df['delivery_days'].fillna(df['delivery_days'].median(), inplace=True)
```

```python
# Type: Categorical: mode
categorical_cols = ['brand', 'payment_method', 'region', 'customer_age_group', 'device_type']
for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
print("Missing values:\n", df.isnull().sum())
```

```python
# Removing duplicates
# ------------------------#
df.drop_duplicates(inplace=True)
```

```python
df['expected_final_price'] = df['price'] * (1 - df['discount_percent'] / 100) * df['quantity']
```

```python
price_mismatch = df[abs(df['final_price'] - df['expected_final_price']) > 1e-2]
print(f"Price mismatches found: {price_mismatch.shape[0]} rows")
```

```python
# Handling mismatches #
```

```python
tolerance = 0.01 * df['expected_final_price']  # 1% tolerance
df.loc[(abs(df['final_price'] - df['expected_final_price']) <= tolerance),
       'final_price'] = df['expected_final_price'].round(2)

df['price_mismatch_flag'] = abs(df['final_price'] - df['expected_final_price']) > tolerance
```

```python
price_mismatch = df[abs(df['final_price'] - df['expected_final_price']) > 1e-2]
print(f"Price mismatches found: {price_mismatch.shape[0]} rows")
```

```python
# Treating outliers (IQR method capping for final_price)
# ------------------------ #
Q1 = df['final_price'].quantile(0.25)
Q3 = df['final_price'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df['final_price'] = df['final_price'].clip(lower, upper)
```

```python
df.boxplot(figsize=(30,30))
plt.show()
```

```python
# Feature engineering (Extract order_month, order_weekday, order_year from order_date.)
# ------------------------#
df['order_month'] = df['order_date'].dt.month
df['order_weekday'] = df['order_date'].dt.day_name()
df['order_year'] = df['order_date'].dt.year
```

```python
# Computing discount_amount = price * discount_percent / 100 & Creating unit_price = final_price / quantity.
```

```python
df['discount_amount'] = df['price'] * df['discount_percent'] / 100
df['unit_price'] = df['final_price'] / df['quantity']
```

```python
# Flagging fast vs slow delivery
df['delivery_speed'] = df['delivery_days'].apply(lambda x: 'fast' if x <= 5 else 'slow')
```

```python
# Customer-level aggregates
# ------------------------#
#Customer-level aggregates: total spend, total orders, average rating given, return rate.

customer_agg = df.groupby('customer_id').agg(
    total_spend=('final_price', 'sum'),
    total_orders=('order_id', 'count'),
    avg_rating=('review_rating', 'mean'),
    return_rate=('is_returned', 'mean')
).reset_index()
```

```python
# Saving cleaned data
# ------------------------#
df.to_csv("cleaned_amazon_sales.csv", index=False)
customer_agg.to_csv("customer_aggregates.csv", index=False)
```

```python
def plot_return_rate(data, category_col):
  return_rate_df = (
        data.groupby(category_col)['is_returned']
        .mean()
        .reset_index()
        .rename(columns={'is_returned': 'return_rate'})
    )
# Convert to percentage
  return_rate_df['return_rate'] = return_rate_df['return_rate'] * 100

# Sort by return rate for better readability
  return_rate_df = return_rate_df.sort_values('return_rate', ascending=False)
# Plot
  plt.figure(figsize=(10, 5))
  sns.barplot(data=return_rate_df, x=category_col, y='return_rate', palette='coolwarm')
  plt.title(f"Return Rate by {category_col}")
  plt.xlabel(category_col)
  plt.ylabel("Return Rate (%)")
  plt.xticks(rotation=45, ha='right')
  plt.show()
```

```python
#-- Run analysis for each required category
# ------------------------ #
plot_return_rate(df, 'main_category')
plot_return_rate(df, 'brand')
plot_return_rate(df, 'customer_age_group')
plot_return_rate(df, 'device_type')
```

```python
# Function to plot return rate for a given categorical column
def plot_return_rate(column):

    # Computing return rate
    return_rate = df.groupby(column)['is_returned'].mean().reset_index()
    return_rate[column] = return_rate[column].astype(str)  # Ensure labels are strings
    return_rate['return_rate'] = return_rate['is_returned'] * 100  # percentage

    # Plotting
    plt.figure(figsize=(10, 5))
    sns.barplot(data=return_rate, x=column, y='return_rate', palette='coolwarm', order=return_rate.sort_values('return_rate', ascending=False)[column])
    plt.title(f"Return Rate by {column}")
    plt.xlabel(column)
    plt.ylabel("Return Rate (%)")
    plt.xticks(rotation=45, ha="right")
    plt.show()
```

```python
df = pd.read_csv("cleaned_amazon_sales.csv")
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

# Snapshot date = day after last order in dataset
snapshot_date = df['order_date'].max() + timedelta(days=1)
```

```python
rfm = df.groupby('customer_id').agg(
    Recency=('order_date', lambda x: (snapshot_date - x.max()).days),
    Frequency=('order_id', 'nunique'),
    Monetary=('final_price', 'sum')
).reset_index()
```

```python
# 3. Behavioral features
# ======================================
# Return rate (% of orders returned)
returns = df.groupby('customer_id')['is_returned'].mean().reset_index()
returns.rename(columns={'is_returned': 'ReturnRate'}, inplace=True)

# Average rating
ratings = df.groupby('customer_id')['review_rating'].mean().reset_index()
ratings.rename(columns={'review_rating': 'AvgRating'}, inplace=True)

# Merge with RFM
rfm = rfm.merge(returns, on='customer_id', how='left') \
         .merge(ratings, on='customer_id', how='left')
```

```python
# 4. Scale features for clustering
# ======================================
features = ['Recency', 'Frequency', 'Monetary', 'ReturnRate', 'AvgRating']
scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm[features])
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
rfm['Cluster'] = kmeans.fit_predict(rfm_scaled)
```

```python
# 6. Label segments (optional)
# ======================================
# Example rules — adjust after inspecting data
segment_labels = {
    0: 'Bargain Hunter',
    1: 'VIP',
    2: 'At Risk',
    3: 'Loyal'
}
rfm['Segment'] = rfm['Cluster'].map(segment_labels)
```

```python
# 7. Segment summary table
# ======================================
segment_summary = rfm.groupby('Segment')[features].mean().reset_index()

print("\nSegment Summary Table:")
print(segment_summary)
```

```python
# 1. Load & Prepare Data
# -------------------------
df = pd.read_csv("cleaned_amazon_sales.csv")

# Select useful numeric features
features = [
    'price', 'discount_percent', 'final_price',
    'quantity', 'delivery_days', 'review_rating'
]
```

```python
# Optionally encode categorical features
categorical = ['main_category', 'sub_category', 'brand', 'payment_method', 'region', 'customer_age_group', 'device_type']
df_encoded = pd.get_dummies(df[features + categorical], drop_first=True)

# Target variable (0 = not returned, 1 = returned)
y = df['is_returned']
X = df_encoded

# -------------------------
# 2. Train / Test Split
# -------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
# 3. Logistic Regression
# -------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_scaled, y_train)

y_pred_lr = log_reg.predict(X_test_scaled)
print("Logistic Regression Report:\n", classification_report(y_test, y_pred_lr))
```

```python
# Logistic Regression Feature Importance
lr_importance = pd.Series(log_reg.coef_[0], index=X.columns).sort_values(key=abs, ascending=False).head(15)

plt.figure(figsize=(8, 6))
sns.barplot(x=lr_importance.values, y=lr_importance.index)
plt.title("Top 15 Logistic Regression Coefficients")
plt.xlabel("Coefficient Value")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
# 4. Decision Tree Classifier
# -------------------------
tree_clf = DecisionTreeClassifier(max_depth=5, random_state=42)
tree_clf.fit(X_train, y_train)

y_pred_tree = tree_clf.predict(X_test)
print("Decision Tree Report:\n", classification_report(y_test, y_pred_tree))

# Decision Tree Feature Importance
tree_importance = pd.Series(tree_clf.feature_importances_, index=X.columns).sort_values(ascending=False).head(15)

plt.figure(figsize=(8, 6))
sns.barplot(x=tree_importance.values, y=tree_importance.index)
plt.title("Top 15 Decision Tree Feature Importances")
plt.xlabel("Importance Score")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
print("Insights:")
print("- Logistic regression tells you the direction & magnitude of influence (positive = increases return likelihood).")
print("- Decision tree shows which features are most useful for splitting customers into returned/non-returned groups.")
print("- Compare both models: If they agree on top features (e.g., delivery_days, discount_percent), these are strong predictors.")
```
